## 4. DATABASE SCHEMA

### 4.1 Entity Relationship Summary

```
Users ──< UserSubjects >── Subjects
Users ──< Schedules
Users ──< Matches >── Users        (self-join)
Users ──< GroupMembers >── StudyGroups
Users ──< Messages
Users ──< Notifications
Users ──< Feedback
Profiles ── Users                   (1:1)
```

### 4.2 Full Table Definitions

#### `users`
```sql
CREATE TABLE users (
  id            UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  email         VARCHAR(255) UNIQUE NOT NULL,
  password_hash VARCHAR(255) NOT NULL,
  full_name     VARCHAR(100) NOT NULL,
  university    VARCHAR(150),
  major         VARCHAR(100),
  year_of_study SMALLINT CHECK (year_of_study BETWEEN 1 AND 6),
  gpa           DECIMAL(3,2) CHECK (gpa BETWEEN 0.0 AND 4.0),
  role          VARCHAR(20) DEFAULT 'student' CHECK (role IN ('student','admin')),
  is_active     BOOLEAN DEFAULT TRUE,
  is_verified   BOOLEAN DEFAULT FALSE,
  last_seen     TIMESTAMP,
  created_at    TIMESTAMP DEFAULT NOW(),
  updated_at    TIMESTAMP DEFAULT NOW()
);
CREATE INDEX idx_users_email ON users(email);
CREATE INDEX idx_users_university ON users(university);
```

#### `profiles`
```sql
CREATE TABLE profiles (
  id              UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  user_id         UUID UNIQUE NOT NULL REFERENCES users(id) ON DELETE CASCADE,
  avatar_url      VARCHAR(500),
  bio             TEXT,
  learning_style  VARCHAR(30) CHECK (learning_style IN (
                    'visual','auditory','reading','kinesthetic','social','solitary')),
  academic_goal   VARCHAR(50) CHECK (academic_goal IN (
                    'exam_prep','certification','job_prep','research','study_abroad','startup')),
  availability    VARCHAR(20) DEFAULT 'flexible' CHECK (availability IN (
                    'morning','afternoon','evening','flexible')),
  xp_points       INTEGER DEFAULT 0,
  streak_days     INTEGER DEFAULT 0,
  is_public       BOOLEAN DEFAULT TRUE,
  created_at      TIMESTAMP DEFAULT NOW(),
  updated_at      TIMESTAMP DEFAULT NOW()
);
```

#### `subjects`
```sql
CREATE TABLE subjects (
  id          UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  name        VARCHAR(100) UNIQUE NOT NULL,    -- e.g. 'Toán cao cấp'
  category    VARCHAR(50),                     -- e.g. 'STEM', 'Business', 'Language'
  code        VARCHAR(20) UNIQUE               -- e.g. 'MATH101'
);
```

#### `user_subjects`
```sql
CREATE TABLE user_subjects (
  user_id     UUID NOT NULL REFERENCES users(id) ON DELETE CASCADE,
  subject_id  UUID NOT NULL REFERENCES subjects(id) ON DELETE CASCADE,
  skill_level VARCHAR(20) DEFAULT 'intermediate' CHECK (skill_level IN (
                'beginner','intermediate','advanced','expert')),
  is_seeking  BOOLEAN DEFAULT TRUE,            -- TRUE=wants help, FALSE=can teach
  PRIMARY KEY (user_id, subject_id)
);
CREATE INDEX idx_user_subjects_user ON user_subjects(user_id);
CREATE INDEX idx_user_subjects_subject ON user_subjects(subject_id);
```

#### `schedules`
```sql
CREATE TABLE schedules (
  id          UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  user_id     UUID NOT NULL REFERENCES users(id) ON DELETE CASCADE,
  day_of_week SMALLINT NOT NULL CHECK (day_of_week BETWEEN 0 AND 6), -- 0=Mon
  time_slot   VARCHAR(10) NOT NULL,            -- e.g. '19-21h'
  UNIQUE (user_id, day_of_week, time_slot)
);
CREATE INDEX idx_schedules_user ON schedules(user_id);
```

#### `matches`
```sql
CREATE TABLE matches (
  id              UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  user_id_a       UUID NOT NULL REFERENCES users(id) ON DELETE CASCADE,
  user_id_b       UUID NOT NULL REFERENCES users(id) ON DELETE CASCADE,
  score           DECIMAL(5,2) NOT NULL,       -- 0.00 to 100.00
  subject_score   DECIMAL(5,2),
  schedule_score  DECIMAL(5,2),
  style_score     DECIMAL(5,2),
  goal_score      DECIMAL(5,2),
  gpa_score       DECIMAL(5,2),
  status          VARCHAR(20) DEFAULT 'suggested' CHECK (status IN (
                    'suggested','pending','connected','rejected')),
  initiated_by    UUID REFERENCES users(id),
  created_at      TIMESTAMP DEFAULT NOW(),
  updated_at      TIMESTAMP DEFAULT NOW(),
  CHECK (user_id_a < user_id_b)               -- prevent duplicate pairs
);
CREATE INDEX idx_matches_user_a ON matches(user_id_a);
CREATE INDEX idx_matches_user_b ON matches(user_id_b);
CREATE INDEX idx_matches_score ON matches(score DESC);
```

#### `study_groups`
```sql
CREATE TABLE study_groups (
  id          UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  name        VARCHAR(100) NOT NULL,
  description TEXT,
  icon        VARCHAR(10),                     -- emoji
  subject_id  UUID REFERENCES subjects(id),
  owner_id    UUID NOT NULL REFERENCES users(id),
  max_members SMALLINT DEFAULT 10,
  next_session TIMESTAMP,
  is_public   BOOLEAN DEFAULT TRUE,
  created_at  TIMESTAMP DEFAULT NOW()
);
```

#### `group_members`
```sql
CREATE TABLE group_members (
  group_id    UUID NOT NULL REFERENCES study_groups(id) ON DELETE CASCADE,
  user_id     UUID NOT NULL REFERENCES users(id) ON DELETE CASCADE,
  role        VARCHAR(20) DEFAULT 'member' CHECK (role IN ('owner','admin','member')),
  joined_at   TIMESTAMP DEFAULT NOW(),
  PRIMARY KEY (group_id, user_id)
);
```

#### `conversations`
```sql
CREATE TABLE conversations (
  id          UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  user_id_a   UUID NOT NULL REFERENCES users(id),
  user_id_b   UUID NOT NULL REFERENCES users(id),
  created_at  TIMESTAMP DEFAULT NOW(),
  updated_at  TIMESTAMP DEFAULT NOW(),
  UNIQUE (user_id_a, user_id_b),
  CHECK (user_id_a < user_id_b)
);
```

#### `messages`
```sql
CREATE TABLE messages (
  id              UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  conversation_id UUID NOT NULL REFERENCES conversations(id) ON DELETE CASCADE,
  sender_id       UUID NOT NULL REFERENCES users(id),
  content         TEXT NOT NULL,
  message_type    VARCHAR(20) DEFAULT 'text' CHECK (message_type IN (
                    'text','image','file','system')),
  is_read         BOOLEAN DEFAULT FALSE,
  created_at      TIMESTAMP DEFAULT NOW()
);
CREATE INDEX idx_messages_conv ON messages(conversation_id, created_at);
```

#### `notifications`
```sql
CREATE TABLE notifications (
  id          UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  user_id     UUID NOT NULL REFERENCES users(id) ON DELETE CASCADE,
  type        VARCHAR(40) NOT NULL,            -- 'match_request','message','group_invite','achievement'
  title       VARCHAR(200) NOT NULL,
  body        TEXT,
  data        JSONB,                           -- flexible payload
  is_read     BOOLEAN DEFAULT FALSE,
  created_at  TIMESTAMP DEFAULT NOW()
);
CREATE INDEX idx_notifs_user ON notifications(user_id, is_read, created_at DESC);
```

#### `feedback`
```sql
CREATE TABLE feedback (
  id              UUID PRIMARY KEY DEFAULT gen_random_uuid(),
  reviewer_id     UUID NOT NULL REFERENCES users(id),
  reviewee_id     UUID NOT NULL REFERENCES users(id),
  match_id        UUID REFERENCES matches(id),
  rating          SMALLINT CHECK (rating BETWEEN 1 AND 5),
  comment         TEXT,
  created_at      TIMESTAMP DEFAULT NOW()
);
```

## 6. AI RECOMMENDATION ENGINE

### Architecture
```
User Request
    │
    ▼
feature_builder.py
    ├── Subject vector (multi-hot encoded, 50-dim)
    ├── Schedule vector (binary 7×6 = 42-dim)
    ├── Learning style (one-hot, 6-dim)
    ├── Academic goal (one-hot, 6-dim)
    └── GPA normalized (1-dim)
                        Total: ~105-dim feature vector
    │
    ▼
preprocessor.py
    └── StandardScaler normalization
    │
    ▼
recommender.py
    ├── KNN (k=20, metric=cosine)
    │   └── Returns top-K candidate users
    ├── Weighted scoring:
    │   ├── subject_overlap × 0.35
    │   ├── schedule_overlap × 0.25
    │   ├── learning_style × 0.15
    │   ├── academic_goal × 0.15
    │   └── gpa_proximity × 0.10
    └── Final score = weighted sum × 100
    │
    ▼
MatchResult(user_id, score, factor_breakdown)
```

### Key Config (ai/recommender.py)
```python
WEIGHTS = {
    "subjects": 0.35,
    "schedule": 0.25,
    "learning_style": 0.15,
    "academic_goal": 0.15,
    "gpa": 0.10,
}
KNN_NEIGHBORS = 20
COSINE_SIMILARITY_THRESHOLD = 0.40
CACHE_TTL_SECONDS = 3600   # re-compute every hour
```

## 9. CODING CONVENTIONS

### Python (Backend)
```
Style guide   : PEP8 + Black formatter (line length 88)
Type hints    : Required on all function signatures
Docstrings    : Google style for all services/AI functions
Error handling: HTTPException with detail dict {code, message}
Logging       : structlog with JSON output
Testing       : pytest + httpx AsyncClient
Naming        :
  - Files     : snake_case.py
  - Classes   : PascalCase
  - Functions : snake_case
  - Constants : UPPER_SNAKE_CASE
  - DB models : User, Profile, Match (singular PascalCase)
```

### TypeScript (Frontend)
```
Style guide   : ESLint + Prettier (2 spaces, single quotes)
Framework     : React 18 functional components only (no class components)
State         : Redux Toolkit (global) + useState/useReducer (local)
Data fetching : React Query (TanStack) + Axios
Naming        :
  - Components: PascalCase (e.g. MatchCard.tsx)
  - Hooks     : camelCase prefixed with use (e.g. useMatches.ts)
  - Services  : camelCase with .service.ts suffix
  - Types     : PascalCase interfaces (e.g. IUser, IMatch)
CSS           : TailwindCSS utility classes
               Custom classes for CócStudy design tokens in globals.css
```

### Database
```
UUIDs         : gen_random_uuid() for all primary keys
Timestamps    : created_at + updated_at on all tables
Soft delete   : is_active flag on users (no hard deletes)
Indexes       : On all FK columns + common query columns
Migrations    : Alembic (one migration per feature set)
Naming        :
  - Tables    : plural snake_case (users, study_groups)
  - Columns   : snake_case
  - Indexes   : idx_{table}_{column}
  - FKs       : fk_{table}_{ref_table}
```

### Git Conventions
```
Branch naming : feature/*, fix/*, refactor/*, docs/*
Commit format : feat: add KNN recommender engine
                fix: correct schedule overlap calculation
                refactor: extract feature builder module
                docs: update API endpoint reference
```
